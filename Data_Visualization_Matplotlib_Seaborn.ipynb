##7. Data Visualization

In this step, we will create four visualizations to understand the Tips dataset, including a histogram, box plot, scatter plot, and correlation heatmap.

####1. Histogram - Distribution of Total Bills

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='total_bill', bins=20, kde=True, color='teal')

plt.title('Distribution of Restaurant Total Bills')
plt.xlabel('Total Bill ($)')
plt.ylabel('Number of Customers')

plt.tight_layout()
plt.show()
```

The histogram shows that most restaurant bills are concentrated in the lower price range, while fewer customers have higher bills.

#### 2. Box Plot - Tips Distribution Across Different Days

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='day', y='tip', color='skyblue')

plt.title('Tip Amounts Across Different Days')
plt.xlabel('Day')
plt.ylabel('Tip Amount ($)')

plt.tight_layout()
plt.show()
```

The box plot shows that tip amounts vary across different days, with some days having higher median tips and more variation than others.

#### 3. Scatter Plot - Relationship Between Total Bill and Tip

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip',
    hue='time',
    palette='Set2',
    s=70
)

plt.title('Higher Restaurant Bills Generally Have Higher Tips')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip Amount ($)')
plt.legend(title='Meal Time')

plt.tight_layout()
plt.show()
```

The scatter plot shows a positive relationship between total bill and tip amount. Customers generally give higher tips when their total bills are higher.

#### 4. Correlation Heatmap - Relationships Between Numerical Variables

```python
correlation = df.select_dtypes(include='number').corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title('Correlation Between Numerical Variables')

plt.tight_layout()
plt.show()
```

The heatmap shows that total_bill and tip have the strongest positive correlation, indicating that higher bills are generally associated with higher tips.

#### 5. Bar Chart - Average Tip by Day

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x='day',
    y='tip',
    hue='day',
    palette='Set2',
    errorbar=None,
    legend=False
)

plt.title('Average Tip Amount by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip ($)')

plt.tight_layout()
plt.show()
```

The bar chart shows that average tip amounts differ across the four days, helping us identify which days have higher average tips.

#### 6. Scatter Plot - Relationship Between Group Size and Tip

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='size',
    y='tip',
    hue='smoker',
    palette='Set1',
    s=80
)

plt.title('Relationship Between Group Size and Tip Amount')
plt.xlabel('Number of People')
plt.ylabel('Tip Amount ($)')
plt.legend(title='Smoker')

plt.tight_layout()
plt.show()
```

The scatter plot shows how tip amounts vary with group size. Larger groups do not always give higher tips.

#### Final 2x2 Visualization Dashboard

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Histogram
sns.histplot(
    data=df,
    x='total_bill',
    bins=20,
    kde=True,
    color='teal',
    ax=axes[0, 0]
)
axes[0, 0].set_title('Most Restaurant Bills Are in the Lower Range')
axes[0, 0].set_xlabel('Total Bill ($)')
axes[0, 0].set_ylabel('Number of Customers')

# 2. Box Plot
sns.boxplot(
    data=df,
    x='day',
    y='tip',
    color='skyblue',
    ax=axes[0, 1]
)
axes[0, 1].set_title('Tip Amounts Vary Across Different Days')
axes[0, 1].set_xlabel('Day')
axes[0, 1].set_ylabel('Tip Amount ($)')

# 3. Scatter Plot
sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip',
    hue='time',
    palette='colorblind',
    s=60,
    ax=axes[1, 0]
)
axes[1, 0].set_title('Higher Bills Generally Have Higher Tips')
axes[1, 0].set_xlabel('Total Bill ($)')
axes[1, 0].set_ylabel('Tip Amount ($)')

# 4. Correlation Heatmap
correlation = df.select_dtypes(include='number').corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    ax=axes[1, 1]
)
axes[1, 1].set_title('Correlation Between Numerical Variables')

# Adjust layout
plt.tight_layout()

# Save dashboard image at 150 DPI
plt.savefig(
    'tips_visualization_dashboard.png',
    dpi=150,
    bbox_inches='tight'
)

plt.show()

print("Dashboard saved successfully at 150 DPI!")
```
